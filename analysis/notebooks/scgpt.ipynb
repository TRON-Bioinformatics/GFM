# scGPT implementation
To run this analysis, we created a new conda environment (specified in analysis/scripts/scgpt_env.yaml)

# Training & Prediction

```python
%%bash
datasets=('replogle_k562' 'replogle_rpe1' 'nadig_hepg2' 'nadig_jurkat' 'norman_combo')
splits=(1 2 3)
for i in "${!datasets[@]}"; do
    dataset="${datasets[$i]}"
    for split in "${splits[@]}"; do
        seed="${split}_0.75"
        split_path="../data/gears/${dataset}/splits/${dataset}_single_${seed}.pkl"
        pert_data_dir="../data/gears/${dataset}"
        save_dir="../data/scgpt/${dataset}/"
        load_model="../data/scgpt/pretrained_model_whole-human"

        python ../scripts/scgpt_train.py ${pert_data_dir} ${split_path} ${split} ${save_dir} ${load_model}
    done
done
```
