To run this analysis, we created a new conda environment (specified in analysis/scripts/gears_env.yaml)

# PertData setup

```python
import scanpy as sc
from gears import PertData
```

```python
datasets = ['replogle_k562', 'replogle_rpe1', 'nadig_hepg2', 'nadig_jurkat']
for dataset in datasets:
    pert_data_dir = '../data/gears/'
    adata = sc.read_h5ad(f'../data/preprocessed_{dataset}.h5ad')
    pert_data = PertData(data_path=pert_data_dir)
    pert_data.new_data_process(dataset_name = dataset, adata = adata)
```

```python
dataset = 'norman_combo'
pert_data_dir = '../data/gears/'
adata = sc.read_h5ad(f'../data/preprocessed_{dataset}.h5ad')
pert_data = PertData(data_path=pert_data_dir)
pert_data.new_data_process(dataset_name = dataset, adata = adata)
# for two-gene perturbation, use gears function to prepare data splits
for seed in [1, 2, 3]:
    pert_data.prepare_split(split='simulation', seed = seed)
```

# Training & Prediction

```python
%%bash
datasets=('replogle_k562' 'replogle_rpe1' 'nadig_hepg2' 'nadig_jurkat' 'norman_combo')
splits=(1 2 3)
for i in "${!datasets[@]}"; do
    dataset="${datasets[$i]}"
    for split in "${splits[@]}"; do
        seed="${split}_0.75"
        split_path="../data/gears/${dataset}/splits/${dataset}_single_${seed}.pkl"
        pert_data_dir="../data/gears/${dataset}"
        save_dir="../data/gears/${dataset}/"

        python ../scripts/gears_train.py ${pert_data_dir} ${seed} ${split_path} ${save_dir}
    done
done
```
